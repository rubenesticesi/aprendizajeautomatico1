# **Validación y Evaluación:**

````python
# Parte 3: Validación y Evaluación
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
import numpy as np

# Seleccionar las variables relevantes y las variables objetivo
# Basado en la exploración inicial, seleccionaremos algunas columnas numéricas relevantes
# y manejaremos los valores faltantes para este ejemplo.
# NOTA: En un análisis real, la selección de variables y el manejo de faltantes
# serían pasos más detallados y considerados.

# Identificar columnas numéricas y de fecha para exclusión inicial o tratamiento especial
numeric_cols = df_suertes.select_dtypes(include=np.number).columns.tolist()
print("Columnas numéricas:", numeric_cols)
datetime_cols = df_suertes.select_dtypes(include='datetime64[ns]').columns.tolist()
print("Columnas de fecha:", datetime_cols)
object_cols = df_suertes.select_dtypes(include='object').columns.tolist()
print("Columnas de objeto:", object_cols)

````

**Interpretación de Tipos de Columnas en `df_suertes`**

Hemos identificado los siguientes tipos de columnas en el DataFrame `df_suertes`:

*   **Columnas Numéricas:** Estas columnas (`int64`, `float64`) contienen valores numéricos y probablemente representan mediciones cuantitativas como 'Período', 'Hacienda', 'Tenencia', 'Area Neta', 'Dist Km', 'TCH', '%Sac.Caña', y muchas otras. Estas columnas serán fundamentales para nuestros modelos de regresión.

*   **Columnas de Fecha:** Estas columnas (`datetime64[ns]`) contienen información temporal, como 'F.Siembra', 'Ult.Riego', 'F.Ult.Corte' y 'Fec.Madur.'. Las fechas son importantes para entender la línea de tiempo de los eventos y podrían ser transformadas (por ejemplo, extrayendo el año o el mes) para ser utilizadas como características en el modelo.

*   **Columnas de Objeto:** Estas columnas (`object`) contienen datos de tipo texto o categórico, como 'Nombre', 'Zona', 'Suerte', 'Suelo', 'Variedad', etc. Para utilizar estas columnas en un modelo de regresión, será necesario aplicar técnicas de codificación, como One-Hot Encoding, si son categóricas nominales, o Label Encoding si son categóricas ordinales.

Comprender el tipo de dato de cada columna es esencial para seleccionar las técnicas de preprocesamiento adecuadas y preparar los datos para el modelado.

````python
# Excluir las columnas de fecha y columnas con un alto porcentaje de valores faltantes
# Para simplificar, solo consideraremos columnas numéricas con menos del 30% de valores faltantes
threshold = 0.30 * len(df_suertes)
cols_to_consider = df_suertes.dropna(axis=1, thresh=threshold).select_dtypes(include=np.number).columns.tolist()

cols_to_consider
````

**Interpretación de `cols_to_consider`**

La lista `cols_to_consider` contiene los nombres de las columnas numéricas del DataFrame `df_suertes` que hemos seleccionado para utilizarlas como características en nuestros modelos de regresión. Para esta selección inicial, hemos excluido las columnas que tenían un porcentaje significativo de valores faltantes (más del 30%) para simplificar el ejemplo.

Estas columnas representan variables que podrían ser relevantes para predecir 'TCH' y '%Sac.Caña', e incluyen información como el período, hacienda, tenencia, área, distancia, edad del último corte, dosis y semanas de madurante, toneladas del último corte, y diversas métricas de lluvia y fosfato.

Es importante recordar que esta es una selección inicial simple. En un análisis más profundo, se deberían considerar otras técnicas de selección de características y métodos más sofisticados para manejar los valores faltantes en las columnas excluidas si se consideran relevantes para el modelo.

````python
# Eliminar las variables objetivo de la lista de características
if 'TCH' in cols_to_consider:
    cols_to_consider.remove('TCH')
if '%Sac.Caña' in cols_to_consider:
    cols_to_consider.remove('%Sac.Caña')

features = cols_to_consider
print("Columnas seleccionadas como características:", features)
target_TCH = 'TCH'
print("Variable objetivo TCH:", target_TCH)
target_SacCana = '%Sac.Caña'
print("Variable objetivo %Sac.Caña:", target_SacCana)

````

````python
# Manejar valores faltantes en las características y variables objetivo (imputación simple con la media)
# NOTA: Esto es una imputación básica. Métodos más avanzados pueden ser necesarios.
df_suertes_processed = df_suertes[features + [target_TCH, target_SacCana]].copy()
for col in features:
    if df_suertes_processed[col].isnull().sum() > 0:
        mean_val = df_suertes_processed[col].mean()
        df_suertes_processed[col].fillna(mean_val, inplace=True)

df_suertes_processed.dropna(subset=[target_TCH, target_SacCana], inplace=True)
df_suertes_processed.info()
````

**Interpretación de `df_suertes_processed` (Después del Preprocesamiento de Faltantes)**

Después de aplicar una imputación simple a los valores faltantes en las columnas seleccionadas como características y eliminar las filas donde faltaban los valores objetivo ('TCH' o '%Sac.Caña'), hemos obtenido el DataFrame `df_suertes_processed`.

La información mostrada (`df_suertes_processed.info()`) confirma que:

*   Ahora tenemos 20,578 entradas (filas), lo que indica que se eliminaron algunas filas debido a la ausencia de valores en las variables objetivo.
*   Las columnas seleccionadas como características (`features`) y las variables objetivo ('TCH', '%Sac.Caña') ahora tienen 20,578 valores no nulos, lo que significa que los valores faltantes en estas columnas han sido manejados (imputados o eliminados).
*   Los tipos de datos de las columnas numéricas y las variables objetivo se mantienen consistentes.

Este DataFrame `df_suertes_processed` está listo para ser utilizado en el entrenamiento de los modelos de regresión, ya que hemos abordado los valores faltantes en las columnas que vamos a utilizar.

````python
# Dividir los datos en conjuntos de entrenamiento y prueba (holdout 80/20)
X = df_suertes_processed[features]
y_TCH = df_suertes_processed[target_TCH]
y_SacCana = df_suertes_processed[target_SacCana]

X_train, X_test, y_TCH_train, y_TCH_test = train_test_split(X, y_TCH, test_size=0.2, random_state=42)
_, _, y_SacCana_train, y_SacCana_test = train_test_split(X, y_SacCana, test_size=0.2, random_state=42)


````

**Interpretación de la División de Datos (Holdout 80/20)**

En este paso, hemos dividido nuestro conjunto de datos procesado (`df_suertes_processed`) en conjuntos de entrenamiento y prueba utilizando la técnica de *holdout* con una proporción 80/20. Esto significa que:

*   El 80% de los datos se asigna al conjunto de entrenamiento (`X_train`, `y_TCH_train`, `y_SacCana_train`). Este conjunto se utilizará para entrenar nuestros modelos de regresión, es decir, para que los modelos "aprendan" la relación entre las características y las variables objetivo (TCH y %Sac.Caña).
*   El 20% restante de los datos se asigna al conjunto de prueba (`X_test`, `y_TCH_test`, `y_SacCana_test`). Este conjunto se mantendrá separado durante el entrenamiento y se utilizará únicamente para evaluar el rendimiento final de los modelos entrenados.

La división aleatoria (`random_state=42`) asegura que la partición sea reproducible y que los datos en ambos conjuntos sean representativos del conjunto de datos original. Al evaluar el modelo en datos que no ha visto durante el entrenamiento (el conjunto de prueba), obtenemos una estimación más realista de cómo se desempeñará el modelo con datos nuevos e invisibles en el futuro.

Hemos realizado esta división tanto para las características (`X`) como para las dos variables objetivo (`y_TCH` y `y_SacCana`) de manera consistente para asegurar que las características y sus correspondientes valores objetivo permanezcan juntos en los conjuntos de entrenamiento y prueba.

````python
# Entrenar modelo de regresión lineal para TCH
model_TCH = LinearRegression()
model_TCH.fit(X_train, y_TCH_train)
````

**Interpretación del Entrenamiento del Modelo de Regresión Lineal para TCH**

En la celda `Entrenar modelo de regresión lineal para TCH`, hemos entrenado un modelo de regresión lineal simple para predecir la variable objetivo 'TCH' (Toneladas de Caña por Hectárea).

*   Primero, creamos una instancia del modelo de regresión lineal utilizando `LinearRegression()`.
*   Luego, ajustamos el modelo a los datos de entrenamiento utilizando el método `fit()`. Le pasamos las características de entrenamiento (`X_train`) y los valores objetivo de entrenamiento para TCH (`y_TCH_train`). Durante este proceso, el modelo encuentra los coeficientes y el intercepto que mejor se ajustan a los datos de entrenamiento para predecir TCH basándose en las características proporcionadas.

Este modelo entrenado está ahora listo para hacer predicciones sobre nuevos datos, como el conjunto de prueba.

````python
# Predecir en el conjunto de prueba para TCH
y_TCH_pred = model_TCH.predict(X_test)
y_TCH_pred
````

**Interpretación de la Predicción en el Conjunto de Prueba para TCH**

Hemos utilizado el modelo de regresión lineal que entrenamos previamente (`model_TCH`) para hacer predicciones sobre las características en nuestro conjunto de prueba (`X_test`).

*   El método `predict()` toma las características del conjunto de prueba como entrada y utiliza los coeficientes y el intercepto que el modelo aprendió durante el entrenamiento para generar valores predichos para 'TCH'.

El resultado, almacenado en la variable `y_TCH_pred`, es un arreglo que contiene las predicciones de TCH para cada una de las filas en el conjunto de prueba. Estos valores predichos se compararán con los valores reales de TCH en el conjunto de prueba (`y_TCH_test`) en el siguiente paso para evaluar el rendimiento del modelo.

````python
# Evaluar el modelo de TCH
r2_TCH = r2_score(y_TCH_test, y_TCH_pred)
rmse_TCH = np.sqrt(mean_squared_error(y_TCH_test, y_TCH_pred))
mae_TCH = mean_absolute_error(y_TCH_test, y_TCH_pred)

print(f"Evaluación del Modelo de Regresión Lineal para TCH:")
print(f"  R²: {r2_TCH:.4f}")
print(f"  RMSE: {rmse_TCH:.4f}")
print(f"  MAE: {mae_TCH:.4f}")

````

**Interpretación de la Evaluación del Modelo de Regresión Lineal para TCH**

hemos evaluado el rendimiento del modelo de regresión lineal entrenado para predecir 'TCH' utilizando el conjunto de prueba. Hemos calculado las siguientes métricas:

*   **R² (Coeficiente de Determinación):** Un valor de R² de 0.9965 indica que el modelo explica aproximadamente el 99.65% de la variabilidad en los valores de TCH en el conjunto de prueba. Un valor cercano a 1 sugiere que el modelo se ajusta muy bien a los datos.

*   **RMSE (Error Cuadrático Medio de la Raíz):** Un RMSE de 1.9511 representa la desviación estándar de los residuos (la diferencia entre los valores predichos y los reales). En promedio, las predicciones de TCH del modelo se desvían de los valores reales en aproximadamente 1.95 unidades. Un RMSE más bajo indica un mejor rendimiento del modelo.

*   **MAE (Error Absoluto Medio):** Un MAE de 1.1164 es el promedio de los errores absolutos entre las predicciones y los valores reales. Al igual que el RMSE, un MAE más bajo indica un mejor rendimiento. El MAE es menos sensible a los outliers que el RMSE.

En general, las métricas de evaluación (R², RMSE, MAE) sugieren que el modelo de regresión lineal tiene un rendimiento muy bueno al predecir TCH en este conjunto de datos. # **Parte 2: Modelo de Clasificación**

## **Creación de Categorías:**

````python
# Parte 2: Modelo de Clasificación
# Creación de Categorías para TCH

# Definir umbrales para TCH usando percentiles
# Usaremos los percentiles 33 y 66 para dividir los datos en tres categorías aproximadamente iguales.
Tch_low_threshold = df_suertes['TCH'].quantile(0.33)
Tch_high_threshold = df_suertes['TCH'].quantile(0.66)

# Crear la nueva columna categórica 'TCH_category'
def categorize_tch(tch):
    if tch <= Tch_low_threshold:
        return 'bajo'
    elif tch <= Tch_high_threshold:
        return 'medio'
    else:
        return 'alto'

df_suertes['TCH_category'] = df_suertes['TCH'].apply(categorize_tch)
# Mostrar el conteo de valores por categoría para verificar la distribución
print("Conteo de valores por categoría de TCH:")
display(df_suertes['TCH_category'].value_counts())
````

````python

# Creación de Categorías para %Sac.Caña

# Definir umbrales para %Sac.Caña usando percentiles
# Usaremos los percentiles 33 y 66 para dividir los datos en tres categorías aproximadamente iguales.
SacCana_low_threshold = df_suertes['%Sac.Caña'].quantile(0.33)
SacCana_high_threshold = df_suertes['%Sac.Caña'].quantile(0.66)

# Crear la nueva columna categórica 'SacCana_category'
def categorize_saccana(saccana):
    if saccana <= SacCana_low_threshold:
        return 'bajo'
    elif saccana <= SacCana_high_threshold:
        return 'medio'
    else:
        return 'alto'

df_suertes['SacCana_category'] = df_suertes['%Sac.Caña'].apply(categorize_saccana)

# Mostrar el conteo de valores por categoría para verificar la distribución
print("\nConteo de valores por categoría de %Sac.Caña:")
display(df_suertes['SacCana_category'].value_counts())
````

**Interpretación de la Creación de Categorías para TCH y %Sac.Caña**

Hemos transformado las variables continuas 'TCH' y '%Sac.Caña' en variables categóricas con tres niveles: 'bajo', 'medio' y 'alto'. La metodología empleada para definir los umbrales entre estas categorías se basa en los **percentiles**:

*   Calculamos el percentil 33 y el percentil 66 para cada variable ('TCH' y '%Sac.Caña').
*   Los valores por debajo o igual al percentil 33 se asignan a la categoría 'bajo'.
*   Los valores entre el percentil 33 (excluido) y el percentil 66 (incluido) se asignan a la categoría 'medio'.
*   Los valores por encima del percentil 66 se asignan a la categoría 'alto'.

Esta metodología basada en percentiles nos permite dividir los datos en tres grupos con un número de observaciones aproximadamente igual en cada categoría (como se puede ver en el conteo de valores). Esto es útil para asegurar que haya suficientes ejemplos en cada clase para entrenar modelos de clasificación de manera efectiva.

La justificación de usar percentiles en este caso es que proporciona una manera sencilla y basada en la distribución de los datos para crear categorías equilibradas, lo cual es un punto de partida común en problemas de clasificación cuando no existen umbrales de negocio predefinidos.

**Justificación de la Validación Cruzada:**

Elegimos la validación cruzada sobre un simple *holdout* (división única en entrenamiento y prueba) por las siguientes razones clave:

1.  **Uso más eficiente de los datos:** La validación cruzada utiliza todo el conjunto de datos tanto para entrenar como para evaluar el modelo. En cada iteración, un subconjunto diferente se reserva para la prueba, lo que nos da una visión más completa de cómo se desempeñaría el modelo en diferentes particiones de los datos. Esto es especialmente valioso en conjuntos de datos de tamaño moderado, donde una única división podría resultar en conjuntos de entrenamiento o prueba no completamente representativos.
2.  **Estimación más robusta del rendimiento:** Al promediar las métricas de evaluación obtenidas de múltiples "folds" (divisiones), la validación cruzada reduce la varianza de la estimación del rendimiento del modelo. Esto nos proporciona una medida más fiable de cómo se espera que el modelo generalice a datos invisibles, en comparación con una única evaluación en un conjunto de prueba particular que podría haber tenido suerte o mala suerte.
3.  **Detección de sobreajuste (Overfitting):** Si el rendimiento del modelo varía significativamente entre los diferentes folds (alta desviación estándar en las métricas), puede indicar que el modelo es sensible a los datos de entrenamiento específicos de cada fold, lo que sugiere un posible sobreajuste. La validación cruzada nos ayuda a identificar esta inestabilidad.
